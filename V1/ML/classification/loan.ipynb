Read the loan CSV into a DataFrame. The target is loan_status: approved or rejected. The pwd command only shows which folder the notebook runs in.

```python
import pandas as pd
```

```python
data = pd.read_csv("../data/loan_data.csv")
```

Check the shape, column types, first rows and basic statistics. This is how you find problems before training.

**3. Clean the column names**
   
The column names have extra spaces at the start, such as ' loan_status'. strip() removes them so the names are easy to use.

```python
data.columns = [i.strip() for i in data.columns]
```

```python
data.columns
```

---
**4. Check unique values and drop the ID**

loan_id is just a row number and has no meaning for the decision. Remove it so the model does not learn from it.

```python
data.nunique
```

```python
data.nunique()
```

```python
data = data.drop(columns=['loan_id'])  # drop a column
```

```python
data.shape  # columns reduced to 12.  It was 13 when we started
```

```python
data.head(4)  # Notice the loan_id removed
```

---
**5. Remove duplicate rows**
   
Count the duplicate rows and drop them. Duplicates can make the model look better than it is.

```python
data.duplicated().sum()  # give me the sum of duplicated data
```

```python
data = data.drop_duplicates(keep='first', ignore_index=False)
```

```python
data.head(4)  # Now will show without a duplication
```

---
**6. Clean the text values** 

Values like ' Yes' and 'Yes' look different to a computer because of the extra space. strip() fixes education, self_employed and loan_status.

```python
data['self_employed'].unique()
```

```python
data['education'] = data['education'].str.strip()
data['self_employed'] = data['self_employed'].str.strip()
data['loan_status'] = data['loan_status'].str.strip()
```

```python
data['self_employed'].unique()
```

---
**7. Handle missing values**
   
Count the empty cells in each column. Fill the empty credit_score values with the column average.

```python
data.isnull().sum()
```

```python
data['credit_score'] = data['credit_score'].fillna(data['credit_score'].mean())  # Add or fill the null value by adding the real value or mean value
```

```python
data.isnull().sum()
```

---
**8. Create a new column: total assets**
   
Add the four asset columns into one total_assets column. Then drop the four originals so the data is smaller and simpler.

```python
data['total_assets'] = data['residential_assets_value'] + data['commercial_assets_value'] + data['luxury_assets_value'] + data['bank_asset_value']
```

```python
data.head(5)
```

```python
data = data.drop(columns=['residential_assets_value', 'commercial_assets_value', 'luxury_assets_value', 'bank_asset_value'])
```

```python
data.head(5)
```

```python
data.shape
```

---
**9. Convert text to numbers [Encoding]**

`Label Encoder`
A model needs numbers, so map each text value to 0 or 1. Graduate=1, Approved=1 and so on.

```python
education_map = {'Graduate': 1, 'Not Graduate': 0}
```

```python
data['education'] = data['education'].map(education_map)
```

```python
data.head(5)
```

```python
self_employed_map = {'No': 1, 'Yes': 0}
data['self_employed'] = data['self_employed'].map(self_employed_map)
```

```python
loan_status_map = {'Approved': 1, 'Rejected': 0}
data['loan_status'] = data['loan_status'].map(loan_status_map)
```

```python
data.head(5)
```

---
**10. Make large values smaller (log)**


Income, loan amount and assets are very large and very uneven. `np.log` brings them to a similar, smaller scale.

```python
import numpy as np
```

```python
data['income_annum'] = np.log(data['income_annum'])
```

```python
data.head()
```

```python
data['loan_amount'] = np.log(data['loan_amount'])
data['total_assets'] = np.log(data['total_assets'])
```

```python
data.head()
```

---
**11. Separate the inputs (X) and the output (Y)**

    
Y is loan_status (0 or 1). X is all the other columns. Check the first rows of both.

```python
X = data.drop(columns=['loan_status'])
```

```python
X.head()
```

```python
Y = data['loan_status']
```

```python
Y.head()
```

---
**12. Split into training and test data**

    
Keep 10% of the rows for testing. stratify=Y keeps the same approved/rejected ratio in both parts.


```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.1, random_state=54, stratify=Y)
```

---
**13. Train a decision tree**

A decision tree learns a set of if/else rules from the training data. fit() builds the tree.

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model = DecisionTreeClassifier()
```

```python
model.fit(x_train, y_train)
```

DecisionTreeClassifier()

---
**14. Predict**
    
predict_proba gives the chance of each class. predict gives the final 0 or 1. Run it on the test data only.

```python
predictions = model.predict(x_test)
```

```python
predictions # 1 = loan approved and 0 = loan rejected
```

---
**15. Accuracy**

Accuracy is the share of test rows predicted correctly. It is a good first number, but not the whole story.

```python
from sklearn.metrics import accuracy_score
```

```python
score = accuracy_score(y_test, predictions)
```

```python
score
```

---
**16. Confusion matrix**

It counts the right and wrong predictions for each class. The table shows exactly which mistakes the model makes.

```python
from sklearn.metrics import confusion_matrix
```

```python
cm = confusion_matrix(y_test, predictions)
```

```python
cm
```

```python
pd.DataFrame(cm, index=['actually Rejected', 'actually Approved'], columns=['predicted Rejected', 'predicted Approved'])
```

joblib writes the trained model to loan.pkl. An API can load it later to make predictions.
